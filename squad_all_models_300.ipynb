```python
import json
import pandas as pd
import sys
import time
from datetime import datetime


sys.path.append("jupyter_dir")
import uq_direction_semantic as uq

# Load SQuAD data (train set)
with open("data/squad/train-v1.1.json", "r") as f:
    squad_data = json.load(f)

# Flatten into question-answer list
qa_list_falcon = []
for article in squad_data["data"]:
    for para in article["paragraphs"]:
        context = para["context"]
        for qa in para["qas"]:
            question = qa["question"]
            # Use the first answer (there may be multiple valid ones)
            answer_text = qa["answers"][0]["text"] if qa["answers"] else None
            qa_list_falcon.append({
                "context": context,
                "question": question,
                "answer_main": answer_text
            })

df_squad = pd.DataFrame(qa_list_falcon)
```

```python
def my_prompt(row, max_ctx_chars=1500):
    """
    Prompt for SQuAD-style QA.
    Uses context when available and truncates to avoid overlong inputs.
    """
    context = row.get("context", None)
    question = row.get("question", "")

    if isinstance(context, str) and context.strip():
        context = context[:max_ctx_chars]
        return (
            "Answer the question using the passage below.\n\n"
            f"Passage: {context}\n\n"
            f"Question: {question}\n"
            "Answer:"
        )

    return f"Question: {question}\nAnswer:"


df_fixed = df_squad.sample(n=30, random_state=40).reset_index(drop=True)
model_ids = [
    "meta-llama/Llama-2-7b-chat-hf",
    "tiiuae/falcon-7b-instruct",
    "mistralai/Mistral-7B-Instruct-v0.2"
]

cache_dir = "model_cache"
hf_token  = "..."


outs = []

for mid in model_ids:
    safe_name = mid.replace("/", "_")
    out_csv = f"squad300_{safe_name}.csv"

    print("\n" + "="*90)
    print(f"START MODEL: {mid}")
    print(f"Output CSV : {out_csv}")
    print(f"Start time : {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")
    t0 = time.time()

    df_out = uq.run_uq(
        df_fixed,
        model_id=mid,
        prompt_fn=my_prompt,
        out_csv=out_csv,
        cache_dir=cache_dir,
        hf_token=hf_token,
        # generation settings (keep identical for fairness)
        n_generations=10,
        batch_size=2,
        max_new_tokens=64,
        temperature=0.7,
        top_p=0.9,
        # entropy baseline
        compute_semantic_entropy=True,
        entail_threshold=0.8,
    )

    t1 = time.time()
    print(f"END MODEL  : {mid}")
    print(f"End time  : {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")
    print(f"Elapsed   : {(t1 - t0)/60:.2f} minutes")
    print("="*90)

    outs.append(df_out)

# -----------------------------
# 7) Combine results across models
# -----------------------------
df_all = pd.concat(outs, ignore_index=True)
out_all_csv = "squad300_all3models_ab30_10samp.csv"
df_all.to_csv(out_all_csv, index=False)
print("\nSaved combined:", out_all_csv)
print("Combined shape:", df_all.shape)
```

```python
import ast
import pandas as pd
import numpy as np

df_all = pd.read_csv("squad300_all3models_ab30_10samp.csv")

def parse_generations(x):
    if isinstance(x, list):
        return x
    if isinstance(x, str):
        x = x.strip()
        if x == "" or x.lower() == "nan":
            return []
        try:
            return ast.literal_eval(x)   # parse "['a','b']" to ['a','b']
        except (SyntaxError, ValueError):
            # fallback: treat as single generation
            return [x]
    return []

df_all["generations"] = df_all["generations"].apply(parse_generations)
```

```python
import numpy as np
from rouge_score import rouge_scorer
from sklearn.metrics import roc_auc_score

# --- use EXACT same scorer ---
scorer = rouge_scorer.RougeScorer(['rougeL'], use_stemmer=True)

# --- EXACT same correctness function, only change is ref column name ---
def is_correct_rouge(row, threshold=0.1):
    generations = row["generations"]
    ref = row["answer_main"]   # TriviaQA reference column

    # Handle missing data
    if not isinstance(ref, str) or not isinstance(generations, list) or len(generations) == 0:
        return False

    top1 = generations[0]
    if not isinstance(top1, str):
        return False

    score = scorer.score(ref.lower(), top1.lower())
    rouge_l_f1 = score['rougeL'].fmeasure
    return rouge_l_f1 >= threshold


# df_all should be your combined TriviaQA results with 3 models
# Example: df_all = pd.read_csv("trivia300_all3models.csv")
# Make sure generations is a list already (in-memory df) OR parsed if loaded from CSV.

THRESH = 0.1

for model_name, df_results in df_all.groupby("model_id"):
    df_results = df_results.copy()

    # --- your exact pipeline ---
    df_results["is_correct"] = df_results.apply(lambda r: is_correct_rouge(r, threshold=THRESH), axis=1)
    df_results["y_true"] = (~df_results["is_correct"]).astype(int)  # 1 = incorrect, 0 = correct

    y_true = df_results["y_true"]

    # AUROC needs both classes
    if y_true.nunique() < 2:
        print(f"\nModel: {model_name}")
        print("Only one class present in y_true, AUROC undefined.")
        print("is_correct mean:", df_results["is_correct"].mean())
        continue

    auroc_kappa = roc_auc_score(y_true, 1 / df_results["kappa"])
    auroc_deberta = roc_auc_score(y_true, df_results["semantic_entropy"])

    print(f"\nModel: {model_name}")
    print(f"AUROC (1/kappa):         {auroc_kappa:.3f}")
    print(f"AUROC (deberta_entropy): {auroc_deberta:.3f}")
```

```python
###Bootstrap
def compute_aurocs(df, threshold=0.1):
    rows = []
    for model_name, df_model in df.groupby("model_id"):
        df_model = df_model.copy()

        # compute correctness
        df_model["is_correct"] = df_model.apply(
            lambda r: is_correct_rouge(r, threshold=threshold),
            axis=1
        )
        df_model["y_true"] = (~df_model["is_correct"]).astype(int)  # 1=incorrect

        y_true = df_model["y_true"]

        # skip if only one class
        if y_true.nunique() < 2:
            rows.append({
                "model_id": model_name,
                "auroc_kappa": np.nan,
                "auroc_semantic": np.nan
            })
            continue

        auroc_kappa = roc_auc_score(y_true, 1 / df_model["kappa"])
        auroc_semantic = roc_auc_score(y_true, df_model["semantic_entropy"])

        rows.append({
            "model_id": model_name,
            "auroc_kappa": auroc_kappa,
            "auroc_semantic": auroc_semantic
        })

    return rows
```

```python
B = 1000
results = []

N = len(df_all)

for b in range(B):
    sample_idx = np.random.choice(N, size=N, replace=True)
    df_boot = df_all.iloc[sample_idx]

    rows = compute_aurocs(df_boot)
    for r in rows:
        r["bootstrap_id"] = b
        results.append(r)

df_bootstrap = pd.DataFrame(results)
#df_bootstrap.to_csv("squad300_auroc_all3models.csv", index = False)
```

```python
summary = (
    df_bootstrap
    .groupby("model_id")
    .agg(
        mean_kappa=("auroc_kappa", "mean"),
        std_kappa=("auroc_kappa", "std"),
        ci95_kappa=("auroc_kappa", lambda x: 1.96 * x.std()),

        mean_sem=("auroc_semantic", "mean"),
        std_sem=("auroc_semantic", "std"),
        ci95_sem=("auroc_semantic", lambda x: 1.96 * x.std()),
    )
)
```

```python
import pandas as pd
import numpy as np
import re

# Step 1: Normalize the kappa to get vmf uncertainty
df_all["vmf_uncert"] = 1.0 / df_all["kappa"]

# Step 2: Function to normalize text for comparison
def norm(s):
    if not isinstance(s, str):
        return ""
    s = s.lower().strip()
    s = re.sub(r"[^\w\s]", "", s)  # remove punctuation
    s = re.sub(r"\s+", " ", s)  # normalize whitespace
    return s

# Step 3: Find the majority answer for each row
def majority_answer(gens):
    return pd.Series([norm(g) for g in gens]).mode()[0]

df_all["maj"] = df_all["generations"].apply(majority_answer)
df_all["truth"] = df_all["answer_main"].apply(norm)

# Check correctness by seeing if the normalized truth is a substring of the majority answer (or vice versa)
df_all["is_correct"] = df_all.apply(
    lambda r: r["truth"] in r["maj"] or r["maj"] in r["truth"], axis=1
)

# Step 4: Define 30% thresholds for low entropy and high vmf uncertainty
se_low_thr  = df_all["semantic_entropy"].quantile(0.60)   # bottom 30%
vmf_high_thr = df_all["vmf_uncert"].quantile(0.30)  # top 30%
vmf_low_thr = df_all["vmf_uncert"].quantile(0.30)

# Step 5: Filter rows that meet the criteria
cases = df_all[
    (df_all["is_correct"] == False) &                     # Model is wrong
    (df_all["semantic_entropy"] <= se_low_thr) &    # Entropy is low (false confidence)
    (df_all["vmf_uncert"] >= vmf_high_thr)    # 1/kappa is high (correctly uncertain)
].copy()

both_fail_cases = df_all[
    (df_all["is_correct"] == False) & 
    (df_all["semantic_entropy"] <= se_low_thr) & 
    (df_all["vmf_uncert"] <= vmf_low_thr)
].copy()

# Step 6: Display a few examples for review
for _, r in both_fail_cases.head(5).iterrows():
    print("="*80)
    print(f"MODEL: {r['model_id']} | ROW: {r['row_id']}")
    print(f"QUESTION: {r['question']}")
    print(f"TRUE ANSWER: {r['answer_main']}")
    print(f"MAJORITY PRED: {r['maj']}")
    print(f"Semantic Entropy: {r['semantic_entropy']:.4f} (low)")
    print(f"1/kappa (vmf_uncert): {r['vmf_uncert']:.4f} (high)")
    print("GENERATIONS:")
    for g in r["generations"]:
        print(f"  - {g}")
    print()
```
